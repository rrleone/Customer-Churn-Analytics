## Análise de Churn por Tipo de Serviço de Internet

O tipo de serviço de internet apresenta diferenças relevantes na taxa de churn.

Clientes com serviço Fiber optic apresentam churn de 41,89%, enquanto clientes DSL apresentam 18,59%. Clientes que não possuem serviço de internet apresentam a menor taxa observada, de 7,40%.

O resultado indica uma associação relevante entre o tipo de serviço de internet e o churn. Entretanto, não é possível afirmar causalidade apenas com essa análise, sendo necessário investigar a relação dessa variável com contrato, tempo de relacionamento e mensalidade.

```python
#Clientes sem serviço de segurança online apresentam maior churn?

pd.crosstab(
    df["OnlineSecurity"],
    df["Churn"],
    normalize="index"
) * 100
```

Clientes sem o serviço OnlineSecurity apresentam taxa de churn de 41,77%, enquanto clientes que possuem o serviço apresentam taxa de 14,61%.

O resultado indica uma associação relevante entre a ausência do serviço de segurança online e maior churn. ## Análise de Churn por Suporte Técnico

Clientes sem o serviço TechSupport apresentam taxa de churn de 41,64%, enquanto clientes que possuem o serviço apresentam taxa de 15,17%.

O resultado é semelhante ao observado para OnlineSecurity, reforçando a existência de uma associação entre a ausência de serviços adicionais e maior churn.

Entretanto, essas variáveis podem estar relacionadas entre si e com outras características dos clientes, como tipo de internet, contrato e mensalidade. Portanto, os resultados não devem ser interpretados como relações causais isoladas.

```python
# Cliente com contrato Month-to-month + até 12 meses de relacionamento + sem OnlineSecurity + sem TechSupport. (hipótese analítica)

perfil_critico = (
    (df["Contract"] == "Month-to-month") &
    (df["TenureGroup"] == "0-12 meses") &
    (df["OnlineSecurity"] == "No") &
    (df["TechSupport"] == "No")
)

df["PerfilRisco"] = perfil_critico.map({
    True: "Perfil crítico",
    False: "Demais clientes"
})
```

```python
df["PerfilRisco"].value_counts()
```

```python
# Qual é a taxa de churn dos clientes que possuem esse perfil em comparação aos demais?

pd.crosstab(
    df["PerfilRisco"],
    df["Churn"],
    normalize="index"
) * 100
```

```python
# Quantos dos 1.188 clientes desse perfil realmente cancelaram?

pd.crosstab(
    df["PerfilRisco"],
    df["Churn"]
)
```

## Identificação de Perfil Crítico de Churn

A combinação de contrato Month-to-month, relacionamento de 0–12 meses, ausência de OnlineSecurity e ausência de TechSupport foi utilizada para identificar um perfil exploratório de maior risco.

O grupo identificado possui 1.188 clientes, dos quais 766 cancelaram, resultando em uma taxa de churn de 64,48%.

Entre os demais 5.855 clientes, a taxa de churn foi de 18,84%.

Embora esse perfil represente aproximadamente 16,9% da base, ele concentra cerca de 41% dos cancelamentos observados. O resultado sugere um segmento prioritário para investigação e possíveis ações de retenção, sem estabelecer relação causal ou capacidade preditiva.

```python
# A taxa de churn varia de acordo com a forma de pagamento utilizada pelo cliente?

pd.crosstab(
    df["PaymentMethod"],
    df["Churn"],
    normalize="index"
) * 100
```

```python
# Quantos clientes utilizam cada forma de pagamento e quantos cancelaram?

pd.crosstab(
    df["PaymentMethod"],
    df["Churn"]
)
```

## Análise de Churn por Forma de Pagamento

A forma de pagamento apresenta diferenças relevantes na taxa de churn.

Clientes que utilizam Electronic check apresentam churn de 45,29%, significativamente superior às demais formas de pagamento, que apresentam taxas entre 15,24% e 19,11%.

O grupo Electronic check possui 2.365 clientes, dos quais 1.071 cancelaram. Esses cancelamentos representam aproximadamente 57% de todos os cancelamentos observados na base.

O resultado indica uma forte associação entre o uso de Electronic check e churn, tornando esse grupo relevante para investigação de possíveis fatores comportamentais e operacionais relacionados à retenção. Entretanto, não é possível estabelecer causalidade apenas com essa análise.

```python
# Forma de pagamento × Contrato

pd.crosstab(
    df["Contract"],
    df["PaymentMethod"],
    normalize="index"
) * 100
```

```python
# Churn dentro de cada contrato

pd.crosstab(
    [df["Contract"], df["PaymentMethod"]],
    df["Churn"],
    normalize="index"
) * 100
```

```python
# Volume de clientes em cada combinação.

pd.crosstab(
    df["Contract"],
    df["PaymentMethod"]
)
```

```python
# Fatores do perfil crítico com Electronic check

pd.crosstab(
    df["PerfilRisco"],
    df["PaymentMethod"]
)
```

```python
# Perfil crítico + Electronic check

perfil_critico_echeck = (
    (df["PerfilRisco"] == "Perfil crítico") &
    (df["PaymentMethod"] == "Electronic check")
)

pd.crosstab(
    perfil_critico_echeck,
    df["Churn"],
    normalize="index"
) * 100
```

## Segmento de Maior Risco Identificado

A análise combinou o perfil crítico previamente identificado com a forma de pagamento Electronic check.

O grupo formado por clientes com contrato Month-to-month, relacionamento de 0–12 meses, ausência de OnlineSecurity e ausência de TechSupport apresentou 69,75% de churn quando também utilizava Electronic check.

Esse segmento possui 734 clientes, dos quais aproximadamente 512 cancelaram. Os cancelamentos desse grupo representam cerca de 27,4% de todos os cancelamentos observados na base.

O resultado identifica um segmento de alta concentração de churn e pode orientar futuras análises e estratégias de retenção. Entretanto, trata-se de uma segmentação exploratória e não de um modelo preditivo ou evidência de causalidade.

# Principais Indicadores de Churn

Nesta etapa são consolidados os principais indicadores identificados durante a análise exploratória.

Os indicadores serão utilizados posteriormente para comparação no Power BI e para apoiar as recomendações de negócio.

```python
# KPIs

total_clientes = len(df)

total_churn = (df["Churn"] == "Yes").sum()

taxa_churn = (total_churn / total_clientes) * 100

clientes_perfil_critico = (df["PerfilRisco"] == "Perfil crítico").sum()

churn_perfil_critico = (
    df.loc[df["PerfilRisco"] == "Perfil crítico", "Churn"]
    .eq("Yes")
    .mean()
) * 100

clientes_perfil_echeck = perfil_critico_echeck.sum()

churn_perfil_echeck = (
    df.loc[perfil_critico_echeck, "Churn"]
    .eq("Yes")
    .mean()
) * 100

print(f"Total de clientes: {total_clientes:,}")
print(f"Total de cancelamentos: {total_churn:,}")
print(f"Taxa de churn: {taxa_churn:.2f}%")
print(f"Clientes no perfil crítico: {clientes_perfil_critico:,}")
print(f"Churn do perfil crítico: {churn_perfil_critico:.2f}%")
print(f"Clientes no perfil crítico + Electronic check: {clientes_perfil_echeck:,}")
print(f"Churn do perfil crítico + Electronic check: {churn_perfil_echeck:.2f}%")
```

7. Clientes que utilizam Electronic check apresentam churn de 45,29%, superior às demais formas de pagamento.

8. A combinação de contrato Month-to-month, relacionamento de 0–12 meses, ausência de OnlineSecurity e ausência de TechSupport identificou um perfil exploratório com churn de 64,48%.
